## Train/Test Split Validation

The updated encoded dataset contains **154,680 observations**.

- **Training set:** 141,823 observations
- **Test set:** 12,857 observations
- **Training period:** May 2025 through May 2026
- **Test period:** June 2026

The most recent available month, **June 2026**, is used exclusively as the test set, while all earlier available months are used for training. This confirms that the dataset follows the required time-based train/test split and prevents future-period information from being used during model training.

```python
from pathlib import Path
import numpy as np
import pandas as pd

# Updated Week 3 encoded dataset
encoded_path = Path(
    "../Week3/model_data_knn_after_encoding.csv.gz"
)

if not encoded_path.exists():
    raise FileNotFoundError(
        f"File not found: {encoded_path.resolve()}"
    )

# Only load metadata first.
# This is fast and lets us verify the June train/test split.
split_check = pd.read_csv(
    encoded_path,
    usecols=[
        "split",
        "SaleMonth",
        "ClosePrice"
    ]
)

print("Dataset shape:", split_check.shape)

print("\nRows by split:")
print(
    split_check["split"].value_counts()
)

print("\nRows by month:")
print(
    split_check
    .groupby(["split", "SaleMonth"])
    .size()
)

print(
    "\nTest month:",
    split_check.loc[
        split_check["split"] == "test",
        "SaleMonth"
    ].unique()
)
```

## Load Final Feature-Engineered Data

The final feature-engineered matrices generated in Week 6 are loaded for model evaluation.

The training set contains all available observations through May 2026, while June 2026 is reserved as the unseen test set.

```python
from pathlib import Path

import numpy as np
from scipy import sparse


# Final Week 6 feature-engineered data
data_dir = Path("../data/week6_final_model_data")

print("Using data directory:")
print(data_dir.resolve())


X_train_path = data_dir / "X_train_week6.npz"
X_test_path = data_dir / "X_test_week6.npz"
y_train_path = data_dir / "y_train_week6.npy"
y_test_path = data_dir / "y_test_week6.npy"


# Check that all files exist
required_files = [
    X_train_path,
    X_test_path,
    y_train_path,
    y_test_path
]

missing_files = [
    path
    for path in required_files
    if not path.exists()
]

if missing_files:
    raise FileNotFoundError(
        "Missing files:\n"
        + "\n".join(str(path) for path in missing_files)
    )


# Load final feature matrices
X_train = sparse.load_npz(
    X_train_path
).tocsr()

X_test = sparse.load_npz(
    X_test_path
).tocsr()

y_train = np.load(
    y_train_path
)

y_test = np.load(
    y_test_path
)


print("\nX_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```
